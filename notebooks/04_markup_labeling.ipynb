````python
exp_dict = {e["experiment_name"]: e["groups"] for e in exp}

TREATMENT_SUMMARIES = {
    # Hassner (Torture)
    "control_base_case": "none (baseline control scenario)",
    "treatment_modest_intelligence": "low expected informational gain only",
    "treatment_protracted_duration": "extended interrogation duration (one month)",
    "treatment_duration_plus_modest_intelligence": "extended duration plus low information gain",
    
    # Savary Goldsmith (Donation)
    "private_donation_control": "none (private donation control scenario)",
    "public_recognition_treatment": "donor name is publicly recognized",
    "public_recognition_with_social_benefit": "public recognition brings social benefits"
}

# =====================================================================
# 2. ФУНКЦИЯ ТРАНСФОРМАЦИИ ДАТАФРЕЙМА
# =====================================================================

def prepare_analysis_dataframe(raw_df):
    prepared_rows = []
    
    for idx, row in raw_df.iterrows():
        exp_name = row.get('experiment_name')
        group_name = row.get('group_name')
        template_name = row.get('template_name', 'two_step_attitude_expression')
        inc_key = row.get('incentive_key')
        
        group_data = exp_dict.get(exp_name, {}).get(group_name, {})
        vignette_text = group_data.get('vignette', 'Text not found')
        question_text = group_data.get('question', 'Question not found')
        answer_options = group_data.get('response_options_text', 'Options not found')

        # --- 1. SYSTEM PROMPT И USER PROMPT ---
        template = prompts_attitude.get(template_name, prompts_attitude.get('two_step_attitude_expression', {}))
        
        if 'STAGE_1_SYSTEM_PROMPT' in template:
            sys_prompt = template['STAGE_1_SYSTEM_PROMPT']
            user_prompt_raw = template['STAGE_1_USER_PROMPT']
        else:
            sys_prompt = template.get('SYSTEM_PROMPT', '')
            user_prompt_raw = template.get('USER_PROMPT', '')
            
        user_prompt = user_prompt_raw.replace('{VIGNETTE_TEXT}', vignette_text)
        user_prompt = user_prompt.replace('{QUESTION_TEXT}', question_text)
        user_prompt = user_prompt.replace('{ANSWER_OPTIONS}', answer_options)
        
        # --- 2. INCENTIVE ---
        if pd.isna(inc_key) or str(inc_key) == 'nan' or inc_key not in incentive:
            final_incentive = "Not provided"
        else:
            final_incentive = incentive[inc_key]
            
        # --- 3. TREATMENT ---
        group_str = str(group_name)
        short_summary = TREATMENT_SUMMARIES.get(group_str, "Unknown treatment")
        
        if pd.isna(group_name) or 'control' in group_str.lower():
            # Добавлена закрывающая скобка и текст виньетки
            final_treatment = f"Treatment applied: NONE ({short_summary})"
        else:
            # Добавлен текст виньетки
            final_treatment = f"Treatment applied: {short_summary}"
            
        # --- 4. REASONING ---
        reasoning = str(row.get('stage1_reasoning_text', ''))
        if reasoning == 'nan' or reasoning.strip() == '':
            reasoning = str(row.get('reasoning_trace_combined', 'Not provided'))
            
        # =================================================================
        # ГЛАВНОЕ ИЗМЕНЕНИЕ: СОХРАНЯЕМ ВСЕ ИСХОДНЫЕ КОЛОНКИ
        # =================================================================
        # Превращаем исходную строку pandas в обычный словарь (dict)
        row_dict = row.to_dict()
        
        # Добавляем в этот словарь наши 5 новых ключей
        row_dict['System prompt'] = sys_prompt.strip()
        row_dict['User prompt'] = user_prompt.strip()
        row_dict['Incentive'] = final_incentive.strip()
        row_dict['Treatment'] = final_treatment.strip()
        row_dict['REASONING'] = reasoning.strip()
        
        # Добавляем расширенный словарь в список
        prepared_rows.append(row_dict)
        
    # Превращаем список обратно в DataFrame. Теперь там будут ВСЕ колонки.
    return pd.DataFrame(prepared_rows)
````

````python
# ─────────────────────────────────────────────
# Настройки чекпоинтов
# ─────────────────────────────────────────────
import logger
CHECKPOINT_EVERY = 50          # сохранять каждые N строк
CHECKPOINT_DIR   = "checkpoints"  # папка для чекпоинтов


def save_checkpoint(
    original_df: pd.DataFrame,
    analyzed_data: list,
    checkpoint_dir: str,
    current_index: int,
) -> str:
    """
    Сохраняет промежуточный CSV-чекпоинт.

    Возвращает путь к сохранённому файлу.
    """
    os.makedirs(checkpoint_dir, exist_ok=True)

    timestamp   = datetime.now().strftime("%Y%m%d_%H%M%S")
    filename    = f"checkpoint_row{current_index}_{timestamp}.csv"
    filepath    = os.path.join(checkpoint_dir, filename)

    # Собираем только уже обработанные строки
    partial_results_df = pd.DataFrame(analyzed_data)
    partial_original   = original_df.iloc[: len(analyzed_data)].reset_index(drop=True)

    partial_df = pd.concat(
        [partial_original, partial_results_df.reset_index(drop=True)],
        axis=1,
    )
    partial_df.to_csv(filepath, index=False, encoding="utf-8-sig")

    logger.info("Чекпоинт сохранён: %s (%d строк)", filepath, len(analyzed_data))
    return filepath


def cleanup_old_checkpoints(checkpoint_dir: str, keep_last: int = 3):
    """
    Удаляет старые чекпоинты, оставляя только keep_last последних.
    Это не даёт папке разрастись.
    """
    if not os.path.exists(checkpoint_dir):
        return

    files = sorted(
        [
            os.path.join(checkpoint_dir, f)
            for f in os.listdir(checkpoint_dir)
            if f.startswith("checkpoint_") and f.endswith(".csv")
        ],
        key=os.path.getmtime,   # сортируем по времени изменения
    )

    # Удаляем всё кроме последних keep_last файлов
    to_delete = files[:-keep_last] if len(files) > keep_last else []
    for old_file in to_delete:
        try:
            os.remove(old_file)
            logger.info("Удалён старый чекпоинт: %s", old_file)
        except OSError as e:
            logger.warning("Не удалось удалить чекпоинт %s: %s", old_file, e)
````

````python
#!/usr/bin/env python3
# -*- coding: utf-8 -*-
"""
Скрипт для анализа reasoning-trace LLM-респондентов через GPT API.
Полностью корректная обработка Unicode (кириллица и др.).
"""

import sys
import json
import time
import logging
import traceback
from datetime import datetime
from collections import defaultdict
from dataclasses import dataclass, field
from typing import Optional

import pandas as pd
from tqdm import tqdm

# ─────────────────────────────────────────────
# Настройка кодировки
# ─────────────────────────────────────────────
if hasattr(sys.stdout, "reconfigure"):
    sys.stdout.reconfigure(encoding="utf-8")
if hasattr(sys.stderr, "reconfigure"):
    sys.stderr.reconfigure(encoding="utf-8")


# ─────────────────────────────────────────────
# Настройка логгера
# ─────────────────────────────────────────────
def setup_logger(log_file: str = "analysis_errors.log") -> logging.Logger:
    """
    Создаёт логгер, который пишет:
      - WARNING+ в файл (с полным стектрейсом)
      - ERROR+   в stderr
    """
    logger = logging.getLogger("llm_analyzer")
    logger.setLevel(logging.DEBUG)

    fmt = logging.Formatter(
        fmt="%(asctime)s | %(levelname)-8s | %(message)s",
        datefmt="%Y-%m-%d %H:%M:%S",
    )

    # Файловый хендлер — все WARNING и выше
    fh = logging.FileHandler(log_file, encoding="utf-8")
    fh.setLevel(logging.WARNING)
    fh.setFormatter(fmt)

    # Консольный хендлер — только ERROR и выше
    ch = logging.StreamHandler(sys.stderr)
    ch.setLevel(logging.ERROR)
    ch.setFormatter(fmt)

    logger.addHandler(fh)
    logger.addHandler(ch)
    return logger


logger = setup_logger()


# ─────────────────────────────────────────────
# Монитор ошибок
# ─────────────────────────────────────────────
@dataclass
class ErrorMonitor:
    """
    Централизованный трекер ошибок.

    Поля:
        total_rows          — всего обработано строк
        success_count       — успешных запросов
        error_counts        — {тип_ошибки: количество}
        error_log           — [{row, type, message, timestamp}, ...]
        retried_rows        — строки, которые потребовали retry
        skipped_rows        — строки с пустым REASONING
        start_time          — время начала обработки
    """
    total_rows: int = 0
    success_count: int = 0
    error_counts: dict = field(default_factory=lambda: defaultdict(int))
    error_log: list = field(default_factory=list)
    retried_rows: list = field(default_factory=list)
    skipped_rows: list = field(default_factory=list)
    start_time: float = field(default_factory=time.time)

    # ── Классификация ошибок ──────────────────
    ERROR_TYPES = {
        "JSONDecodeError":    "Невалидный JSON в ответе API",
        "RateLimitError":     "Превышен лимит запросов (Rate Limit)",
        "APIConnectionError": "Ошибка соединения с API",
        "AuthenticationError":"Ошибка аутентификации API-ключа",
        "Timeout":            "Таймаут запроса",
        "EmptyReasoning":     "Поле REASONING пустое или отсутствует",
        "UnknownError":       "Неклассифицированная ошибка",
    }

    def _classify(self, error: Exception) -> str:
        """Определяет тип ошибки по классу исключения."""
        name = type(error).__name__
        for key in self.ERROR_TYPES:
            if key.lower() in name.lower():
                return key
        return "UnknownError"

    def record_success(self):
        self.total_rows += 1
        self.success_count += 1

    def record_skip(self, row_index: int, reason: str):
        """Регистрирует пропущенную строку (например, пустой REASONING)."""
        self.total_rows += 1
        self.skipped_rows.append({"row": row_index, "reason": reason})
        self.error_counts["EmptyReasoning"] += 1
        logger.warning("Row %d | SKIP | %s", row_index, reason)

    def record_error(
        self,
        row_index: int,
        error: Exception,
        context: Optional[str] = None,
        retried: bool = False,
    ):
        """
        Регистрирует ошибку:
          - увеличивает счётчик нужного типа
          - сохраняет запись в error_log
          - пишет в лог-файл (WARNING) и stderr (ERROR)
          - сохраняет полный стектрейс в лог-файл
        """
        self.total_rows += 1
        error_type = self._classify(error)
        self.error_counts[error_type] += 1

        entry = {
            "row":       row_index,
            "type":      error_type,
            "message":   repr(error),
            "context":   context or "",
            "retried":   retried,
            "timestamp": datetime.now().isoformat(timespec="seconds"),
        }
        self.error_log.append(entry)

        if retried:
            self.retried_rows.append(row_index)

        # Полный стектрейс — только в файл (WARNING)
        tb = traceback.format_exc()
        logger.warning(
            "Row %d | %s | retried=%s | %s\nTraceback:\n%s",
            row_index, error_type, retried, repr(error), tb,
        )
        # Краткое сообщение — в stderr (ERROR)
        logger.error(
            "Row %d | %s | %s", row_index, error_type, repr(error)
        )

    # ── Финальный отчёт ──────────────────────
    def print_report(self):
        """Выводит итоговый отчёт в stdout и пишет его в лог."""
        elapsed = time.time() - self.start_time
        error_total = sum(self.error_counts.values())

        lines = [
            "",
            "=" * 60,
            "           ОТЧЁТ О МОНИТОРИНГЕ ОШИБОК",
            "=" * 60,
            f"  Время выполнения  : {elapsed:.1f} сек",
            f"  Всего строк       : {self.total_rows}",
            f"  ✅ Успешно         : {self.success_count}",
            f"  ❌ Ошибок          : {error_total}",
            f"  ⏭  Пропущено       : {len(self.skipped_rows)}",
            f"  🔁 Retry-запросов  : {len(self.retried_rows)}",
        ]

        if self.error_counts:
            lines.append("")
            lines.append("  Разбивка по типам ошибок:")
            for etype, cnt in sorted(
                self.error_counts.items(), key=lambda x: -x[1]
            ):
                desc = self.ERROR_TYPES.get(etype, etype)
                lines.append(f"    [{cnt:>4}]  {etype:<25} — {desc}")

        if self.retried_rows:
            sample = self.retried_rows[:10]
            lines.append("")
            lines.append(
                f"  Строки с retry (первые 10): {sample}"
                + (" ..." if len(self.retried_rows) > 10 else "")
            )

        if self.error_log:
            lines.append("")
            lines.append("  Последние 5 ошибок:")
            for entry in self.error_log[-5:]:
                lines.append(
                    f"    • row={entry['row']} | {entry['type']} "
                    f"| {entry['timestamp']}"
                )
                lines.append(f"      {entry['message'][:120]}")

        lines.append("=" * 60)

        report_text = "\n".join(lines)
        print(report_text)
        logger.warning("ФИНАЛЬНЫЙ ОТЧЁТ:\n%s", report_text)

    def save_error_log(self, path: str = "error_log.json"):
        """Сохраняет полный лог ошибок в JSON для постанализа."""
        with open(path, "w", encoding="utf-8") as f:
            json.dump(self.error_log, f, ensure_ascii=False, indent=2)
        print(f"   📄 Лог ошибок : {path}")


# ─────────────────────────────────────────────
# Глобальный монитор (создаётся один раз)
# ─────────────────────────────────────────────
monitor = ErrorMonitor()


# ─────────────────────────────────────────────
# Вспомогательные функции
# ─────────────────────────────────────────────
def safe_text(value) -> str:
    """Нормализует любое значение в безопасную строку UTF-8."""
    if value is None:
        return "Not provided"
    if isinstance(value, bytes):
        try:
            text = value.decode("utf-8")
        except UnicodeDecodeError:
            text = value.decode("utf-8", errors="replace")
    elif isinstance(value, str):
        text = value
    else:
        try:
            if pd.isna(value):
                return "Not provided"
        except (TypeError, ValueError):
            pass
        text = str(value)

    text = text.strip()
    if not text or text.lower() == "nan":
        return "Not provided"
    return text.replace("\x00", " ")


def flatten_json(data: dict) -> dict:
    """Рекурсивно разворачивает вложенный JSON в плоский dict."""
    out = {}

    def _flatten(obj, prefix=""):
        if isinstance(obj, dict):
            for key, val in obj.items():
                _flatten(val, f"{prefix}{key}_")
        elif isinstance(obj, list):
            out[prefix.rstrip("_")] = ", ".join(str(item) for item in obj)
        else:
            out[prefix.rstrip("_")] = obj

    _flatten(data)
    return out


def empty_result_row() -> dict:
    """Пустая структура результата — гарантирует одинаковые колонки."""
    return {
        "treatment_awareness_score": None,
        "treatment_awareness_explanation": None,
        "incentive_awareness_score": None,
        "incentive_awareness_explanation": None,
        "reasoning_style_primary_strategy": None,
        "reasoning_style_internal_conflict": None,
        "epistemic_certainty": None,
        "extracted_topics": None,
        "social_desirability_normative_orientation": None,
        "social_desirability_fear_of_judgment": None,
        "social_desirability_safe_answer_shift": None,
        "role_adherence_ai_leakage": None,
        "role_adherence_role_immersion": None,
        "affective_state_valence": None,
        "affective_state_dominant_emotion": None,
        "cognitive_attribution_locus_of_control": None,
        "cognitive_attribution_temporal_focus": None,
        "epistemic_assumptions_hallucinated_conditions": None,
        "api_error": None,
    }


# ─────────────────────────────────────────────
# Основная функция анализа строки
# ─────────────────────────────────────────────
MAX_RETRIES = 3        # максимум попыток при временных ошибках
RETRY_DELAY = 5        # секунд между попытками

# Ошибки, при которых имеет смысл делать retry
RETRYABLE_ERRORS = ("RateLimitError", "APIConnectionError", "Timeout")


def analyze_row(row, row_index: int = 0) -> dict:
    """
    Отправляет reasoning-trace в GPT и возвращает плоский dict.

    Логика:
      1. Если REASONING пустой — пропускаем (monitor.record_skip)
      2. До MAX_RETRIES попыток при временных ошибках (monitor.record_error retried=True)
      3. При постоянной ошибке — monitor.record_error retried=False
      4. При успехе — monitor.record_success()
    """
    sys_prompt  = safe_text(row.get("System prompt"))
    user_prompt = safe_text(row.get("User prompt"))
    incentive   = safe_text(row.get("Incentive"))
    treatment   = safe_text(row.get("Treatment"))
    reasoning   = safe_text(row.get("REASONING"))

    # ── Пропуск пустого reasoning ────────────
    if reasoning == "Not provided":
        monitor.record_skip(row_index, "REASONING is empty or missing")
        result = empty_result_row()
        result["api_error"] = "REASONING is empty or missing"
        return result

    user_message = (
        "EXPERIMENT CONTEXT:\n"
        f"- System prompt: {sys_prompt}\n"
        f"- User prompt:   {user_prompt}\n"
        f"- Incentive:     {incentive}\n"
        f"- Treatment:     {treatment}\n\n"
        "REASONING TEXT:\n"
        f"{reasoning}"
    )

    last_error: Optional[Exception] = None

    for attempt in range(1, MAX_RETRIES + 1):
        try:
            response = client.chat.completions.create(
                model=MODEL_NAME,
                messages=[
                    {"role": "system", "content": SYSTEM_PROMPT},
                    {"role": "user",   "content": user_message},
                ],
                response_format={"type": "json_object"},
                temperature=0.1,
            )

            raw_json    = response.choices[0].message.content
            parsed_dict = json.loads(raw_json)
            flat        = flatten_json(parsed_dict)
            flat["api_error"] = None

            # ✅ Успех — считаем только здесь
            monitor.record_success()
            logger.debug("Row %d | OK (attempt %d)", row_index, attempt)
            return flat

        except json.JSONDecodeError as e:
            # Невалидный JSON — retry не поможет
            monitor.record_error(row_index, e, context=f"attempt={attempt}", retried=False)
            result = empty_result_row()
            result["api_error"] = f"JSON parse error: {e}"
            return result

        except Exception as e:
            last_error  = e
            error_type  = monitor._classify(e)
            is_retryable = error_type in RETRYABLE_ERRORS

            if is_retryable and attempt < MAX_RETRIES:
                # Временная ошибка — пробуем ещё раз
                monitor.record_error(
                    row_index, e,
                    context=f"attempt={attempt}/{MAX_RETRIES}",
                    retried=True,
                )
                # Не считаем total_rows повторно — убираем один лишний инкремент
                monitor.total_rows -= 1
                logger.warning(
                    "Row %d | retry %d/%d after %.1fs | %s",
                    row_index, attempt, MAX_RETRIES, RETRY_DELAY, repr(e),
                )
                time.sleep(RETRY_DELAY)
            else:
                # Постоянная ошибка или исчерпаны попытки
                monitor.record_error(
                    row_index, e,
                    context=f"final attempt={attempt}",
                    retried=False,
                )
                result = empty_result_row()
                result["api_error"] = repr(e)
                return result

    # Сюда не должны попасть, но на всякий случай
    result = empty_result_row()
    result["api_error"] = repr(last_error)
    return result


# ─────────────────────────────────────────────
# Обработка всего датасета
# ─────────────────────────────────────────────
def process_dataset(input_csv: str, output_csv: str, output_parquet: str):
    """Загружает CSV, прогоняет каждую строку через GPT, сохраняет результат."""

    print(f"Загрузка данных из {input_csv}...")
    df = pd.read_csv(input_csv, encoding="utf-8")

    required_cols = ["System prompt", "User prompt", "Incentive", "Treatment", "REASONING"]
    missing = [col for col in required_cols if col not in df.columns]
    if missing:
        logger.error("Отсутствуют обязательные колонки: %s", missing)
        print(f"⚠️  Отсутствуют колонки: {missing}")
        print(f"   Имеющиеся колонки: {list(df.columns)}")

    analyzed_data       = []
    last_checkpoint_row = None   # путь к последнему чекпоинту

    print(f"Запуск анализа ({MODEL_NAME}), строк: {len(df)}...")
    print(f"📌 Чекпоинт каждые {CHECKPOINT_EVERY} строк → папка '{CHECKPOINT_DIR}/'")

    for idx, row in tqdm(df.iterrows(), total=len(df), desc="Analyzing"):
        result = analyze_row(row, row_index=idx)
        analyzed_data.append(result)

        rows_done = len(analyzed_data)

        # ── Чекпоинт каждые CHECKPOINT_EVERY строк ──
        if rows_done % CHECKPOINT_EVERY == 0:
            last_checkpoint_row = save_checkpoint(
                original_df    = df,
                analyzed_data  = analyzed_data,
                checkpoint_dir = CHECKPOINT_DIR,
                current_index  = idx,
            )
            # Оставляем только 3 последних чекпоинта
            cleanup_old_checkpoints(CHECKPOINT_DIR, keep_last=3)

            tqdm.write(
                f"\n📌 Чекпоинт [{rows_done}/{len(df)}] → {last_checkpoint_row}"
            )

    # ── Финальный чекпоинт (если последний батч < CHECKPOINT_EVERY) ──
    remainder = len(analyzed_data) % CHECKPOINT_EVERY
    if remainder != 0:
        last_checkpoint_row = save_checkpoint(
            original_df    = df,
            analyzed_data  = analyzed_data,
            checkpoint_dir = CHECKPOINT_DIR,
            current_index  = len(df) - 1,
        )
        tqdm.write(f"\n📌 Финальный чекпоинт → {last_checkpoint_row}")

    # ── Итоговые файлы ───────────────────────────
    results_df = pd.DataFrame(analyzed_data)
    final_df   = pd.concat(
        [df.reset_index(drop=True), results_df.reset_index(drop=True)],
        axis=1,
    )

    final_df.to_csv(output_csv, index=False, encoding="utf-8-sig")
    final_df.to_parquet(output_parquet, index=False, engine="pyarrow")

    # ── Отчёт ────────────────────────────────────
    monitor.print_report()
    if monitor.error_log:
        monitor.save_error_log("error_log.json")

    print(f"\nФайлы сохранены:")
    print(f"   CSV:        {output_csv}")
    print(f"   Parquet:    {output_parquet}")
    print(f"   Лог:        analysis_errors.log")
    if last_checkpoint_row:
        print(f"   Чекпоинты: {CHECKPOINT_DIR}/")
    print(f"\nДобавленные колонки:")
    for col in results_df.columns:
        print(f"   • {col}")

````
